```python
import numpy as np
import pandas as pd

# 1 - import dataset in my file
data = pd.read_csv("Dataset.csv")

# 2 - missing value
data.replace(['Not Given', 'UnKnown', 'UnKnown', 'N/A', 'NAN', ''], np.nan, inplace=True)

data.isnull().any() # director , country

data['country'].isnull().sum() # 287
data['director'].isnull().sum() # 2588

data['director'] = data['director'].fillna('Unknown')
data['country'] = data['country'].fillna(data['country'].mode()[0])

# 3 - data dubplicate 
print("the number of rows before delete a data duplicated ", len(data))

data.drop_duplicates(inplace=True)
data.drop_duplicates(subset=['show_id'], keep='first', inplace=True)

print("the number of rows after delete a data duplicated ", len(data))

# 3 - formating 
text_cols = ['title', 'director', 'country', 'rating', 'listed_in', 'type']

for col in text_cols :
    if col in data.columns :
        data[col] = data[col].astype(str).str.strip()

if 'type' in data.columns :
    data['type'] = data['type'].str.title()

if 'date_added' in data.columns:
    data['date_added'] = pd.to_datetime(data['date_added'], errors='coerce')

# 4 - standarde columns and modify Rating column
data.columns = data.columns.str.strip().str.replace(' ', '_').str.title()

data['Rating'] = data['Rating'].str.upper().str.strip()

# 5 - export cleaned data
Modified_data = data.copy()
Modified_data.to_csv("Cleaned_Dataset.csv", index=False)


print("\nThe file has been successfully cleaned and formatted , and exported to a new file named 'Cleaned_Dataset'\n")
```
