```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv("Cleaned_Dataset.csv")
year_trend = data.groupby(['Release_Year', 'Type']).size().unstack(fill_value=0)

# Ploting films and tv shows productions within a single plot
plt.figure(figsize=(14, 7))
sns.lineplot(data=year_trend, linewidth=2.5, palette=['#E50914', '#221F1F'])

plt.title('Netflix Content Production Trend by Release Year (Full Range)', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Release Year', fontsize=12, fontweight='bold')
plt.ylabel('Number of Titles', fontsize=12, fontweight='bold')
plt.legend(title='Content Type', title_fontsize='11', fontsize='10', loc='upper left')
plt.grid(True, linestyle='--', alpha=0.5)

plt.xlim(year_trend.index.min(), year_trend.index.max())

plt.tight_layout()
plt.show()
```

```python
# Plot a movies only

plt.figure(figsize=(14, 10))
plt.subplot(2, 1, 1)

sns.lineplot(data=year_trend['Movie'], color='#E50914', linewidth=2.5, label='Movies')

plt.title('Netflix Movies Production Trend by Release Year', fontsize=14, fontweight='bold', pad=12)
plt.xlabel('Release Year', fontsize=12, fontweight='bold')
plt.ylabel('Number of Movies', fontsize=12, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.5)
plt.xlim(year_trend.index.min(), year_trend.index.max())
plt.legend(loc='upper left', fontsize='10')

plt.tight_layout()
plt.show()
```

```python
# Plot a tv shows only

plt.figure(figsize=(14, 10))
plt.subplot(2, 1, 2)

sns.lineplot(data=year_trend['Tv Show'], color='#221F1F', linewidth=2.5, label='TV Shows')

plt.title('Netflix TV Shows Production Trend by Release Year', fontsize=14, fontweight='bold', pad=12)
plt.xlabel('Release Year', fontsize=12, fontweight='bold')
plt.ylabel('Number of TV Shows', fontsize=12, fontweight='bold')
plt.grid(True, linestyle='--', alpha=0.5)
plt.xlim(year_trend.index.min(), year_trend.index.max())
plt.legend(loc='upper left', fontsize='10')

plt.tight_layout()
plt.show()
```
