```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv('Cleaned_Dataset.csv')

type_counts = df['Type'].value_counts()
type_percentages = df['Type'].value_counts(normalize=True) * 100

plt.style.use('seaborn-v0_8-whitegrid')
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
colors = ['#E50914', '#221F1F'] 

axes[0].pie(
    type_counts,
    labels=type_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    pctdistance=0.78,
    colors=colors,
    wedgeprops=dict(width=0.4, edgecolor='white', linewidth=2),
    textprops=dict(fontsize=12, fontweight='bold', color='white'),
)
axes[0].set_title(
    'Distribution of Content Types on Netflix', fontsize=14, fontweight='bold'
)
axes[0].text(
    0,
    0,
    f'Total\n{len(df):,}\nTitles',
    ha='center',
    va='center',
    fontsize=12,
    fontweight='bold',
)

barplot = sns.barplot(
    x=type_counts.index, y=type_counts.values, ax=axes[1], palette=colors
)
axes[1].set_title(
    'Total Count of Movies vs TV Shows', fontsize=14, fontweight='bold'
)
axes[1].set_ylabel('Number of Titles', fontsize=11, fontweight='bold')
axes[1].set_xlabel('Content Type', fontsize=11, fontweight='bold')

for p in barplot.patches:
  height = p.get_height()
  axes[1].annotate(
      f'{int(height):,}\n({height/len(df)*100:.1f}%)',
      (p.get_x() + p.get_width() / 2.0, height / 2),
      ha='center',
      va='center',
      fontsize=11,
      color='white',
      fontweight='bold',
  )

plt.tight_layout()
plt.show()
```
